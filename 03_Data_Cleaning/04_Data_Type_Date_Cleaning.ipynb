```python
import pandas as pd
import numpy as np

from pathlib import Path

file_path = Path("../01_Raw_Data/marketing_data.xlsx")

df_raw = pd.read_excel(file_path)

df = df_raw.copy()

# Sütun adlarını standartlaşdırırıq
df.columns = df.columns.str.strip()

print("Dataset Shape:", df.shape)

display(df.head())
```

```python
data_types = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values
})

display(data_types)
```

```python
print("Income Data Type:", df["Income"].dtype)

display(df["Income"].head())
```

```python
print("Original Data Type:", df["Dt_Customer"].dtype)

display(df["Dt_Customer"].head())
```

```python
df["Enrollment_Year"] = (
    df["Dt_Customer"].dt.year
)

df["Enrollment_Month"] = (
    df["Dt_Customer"].dt.month
)

df["Enrollment_Quarter"] = (
    df["Dt_Customer"].dt.quarter
)

display(
    df[
        [
            "ID",
            "Dt_Customer",
            "Enrollment_Year",
            "Enrollment_Month",
            "Enrollment_Quarter"
        ]
    ].head()
)
```

```python
df["Enrollment_Month_Name"] = (
    df["Dt_Customer"].dt.month_name()
)
```

```python
print("Data Type:", df["Year_Birth"].dtype)

print("Minimum:", df["Year_Birth"].min())
print("Maximum:", df["Year_Birth"].max())
```

```python
invalid_birth_year = df[
    (df["Year_Birth"] < 1900) |
    (df["Year_Birth"] > 2014)
]

display(
    invalid_birth_year[
        ["ID", "Year_Birth"]
    ]
)
```

```python
reference_year = 2014

df["Customer_Age_Estimated"] = (
    reference_year - df["Year_Birth"]
)
```

```python
display(
    df[
        [
            "ID",
            "Year_Birth",
            "Customer_Age_Estimated"
        ]
    ].sort_values(
        "Customer_Age_Estimated",
        ascending=False
    ).head(10)
)
```

```python
validation_summary = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Unique Customer IDs",
        "Income NULL",
        "Invalid Dates",
        "Invalid Birth Years"
    ],
    "Value": [
        len(df),
        df["ID"].nunique(),
        df["Income"].isna().sum(),
        df["Dt_Customer"].isna().sum(),
        len(invalid_birth_year)
    ]
})

display(validation_summary)
```
