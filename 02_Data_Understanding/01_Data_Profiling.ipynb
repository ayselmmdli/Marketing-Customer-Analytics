```python
# Sütunların data type-ları düzgündürmü?
#display(df.dtypes.to_frame(name="Data_Type"))

#print("Dt_Customer Data Type:")
#print(df["Dt_Customer"].dtype)

#print("\nFirst 5 Dates:")
#display(df["Dt_Customer"].head())

import pandas.api.types as ptypes

type_groups = {
    "integer": [
        "ID", "Year_Birth", "Kidhome", "Teenhome",
        "Recency", "NumDealsPurchases", "NumWebPurchases",
        "NumCatalogPurchases", "NumStorePurchases",
        "NumWebVisitsMonth", "AcceptedCmp1", "AcceptedCmp2",
        "AcceptedCmp3", "AcceptedCmp4", "AcceptedCmp5",
        "Response", "Complain"
    ],
    "numeric": [
        "Income", "MntWines", "MntFruits",
        "MntMeatProducts", "MntFishProducts",
        "MntSweetProducts", "MntGoldProds"
    ],
    "date": ["Dt_Customer"],
    "text": ["Education", "Marital_Status", "Country"]
}

results = []

for group, columns in type_groups.items():
    for col in columns:
        if col not in df.columns:
            results.append([col, group, "Missing", "CHECK"])
            continue

        dtype = df[col].dtype

        if group == "integer":
            valid = ptypes.is_integer_dtype(dtype)
        elif group == "numeric":
            valid = ptypes.is_numeric_dtype(dtype)
        elif group == "date":
            valid = ptypes.is_datetime64_any_dtype(dtype)
        else:
            valid = (
                ptypes.is_object_dtype(dtype)
                or ptypes.is_string_dtype(dtype)
            )

        results.append([
            col,
            group,
            str(dtype),
            "PASS" if valid else "CHECK"
        ])

type_validation = pd.DataFrame(
    results,
    columns=["Column", "Expected", "Actual", "Status"]
)

display(type_validation)
```

```python
# Sütun adlarının əvvəlindəki və sonundakı boşluqları silirik
df.columns = df.columns.str.strip()

# Nəticəni yoxlayırıq
print(df.columns.tolist())
```

```python
numeric_columns = [
    "Income",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds",
    "NumDealsPurchases",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "NumWebVisitsMonth",
    "Recency"
]

negative_values = (df[numeric_columns] < 0).sum()

display(
    negative_values.to_frame(
        name="Negative_Count"
    )
)
```

```python
binary_columns = [
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5",
    "Response",
    "Complain"
]

for col in binary_columns:
    invalid_count = (
        ~df[col].isin([0, 1])
    ).sum()

    print(col, ":", invalid_count)
```

```python
outlier_columns = [
    "Income",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    outlier_results.append({
        "Column": col,
        "Lower_Bound": round(lower, 2),
        "Upper_Bound": round(upper, 2),
        "Outlier_Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```
