```python
import pandas as pd
import numpy as np

from pathlib import Path

file_path = Path("../01_Raw_Data/marketing_data.xlsx")

df_raw = pd.read_excel(file_path)

df = df_raw.copy()

# Sütun adlarının standartlaşdırılması
df.columns = df.columns.str.strip()

# Tarix formatı
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    errors="coerce"
)

# Gəlir formatı
df["Income"] = pd.to_numeric(
    df["Income"],
    errors="coerce"
)

print("Dataset Shape:", df.shape)

display(df.head())
```

```python
spending_columns = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

df["Total_Spending"] = (
    df[spending_columns].sum(axis=1)
)

display(
    df[
        ["ID", "Total_Spending"]
    ].head(10)
)
```

```python
purchase_columns = [
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases"
]

df["Total_Purchases"] = (
    df[purchase_columns].sum(axis=1)
)

display(
    df[
        [
            "ID",
            "NumWebPurchases",
            "NumCatalogPurchases",
            "NumStorePurchases",
            "Total_Purchases"
        ]
    ].head(10)
)
```

```python
df["Total_Children"] = (
    df["Kidhome"] + df["Teenhome"]
)

display(
    df[
        [
            "ID",
            "Kidhome",
            "Teenhome",
            "Total_Children"
        ]
    ].head()
)
```

```python
campaign_columns = [
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5",
    "Response"
]

df["Total_Accepted_Campaigns"] = (
    df[campaign_columns].sum(axis=1)
)

display(
    df[
        [
            "ID",
            "Total_Accepted_Campaigns"
        ]
    ].head(10)
)
```

```python
reference_year = 2014

df["Customer_Age_Estimated"] = (
    reference_year - df["Year_Birth"]
)
```

```python
invalid_age = df[
    (df["Customer_Age_Estimated"] < 18) |
    (df["Customer_Age_Estimated"] > 100)
]

print("Potentially Invalid Ages:", len(invalid_age))

display(
    invalid_age[
        [
            "ID",
            "Year_Birth",
            "Customer_Age_Estimated"
        ]
    ]
)
```

```python
reference_date = pd.Timestamp("2014-06-30")

df["Customer_Tenure_Days"] = (
    reference_date - df["Dt_Customer"]
).dt.days

display(
    df[
        [
            "ID",
            "Dt_Customer",
            "Customer_Tenure_Days"
        ]
    ].head(10)
)
```

```python
age_bins = [18, 25, 35, 45, 55, 65, np.inf]

age_labels = [
    "18-24",
    "25-34",
    "35-44",
    "45-54",
    "55-64",
    "65+"
]

df["Age_Group"] = pd.cut(
    df["Customer_Age_Estimated"],
    bins=age_bins,
    labels=age_labels,
    right=False
)

display(
    df[
        [
            "ID",
            "Customer_Age_Estimated",
            "Age_Group"
        ]
    ].head(10)
)
```

```python
df["Income_Group"] = pd.qcut(
    df["Income"],
    q=4,
    labels=[
        "Low Income",
        "Lower-Middle Income",
        "Upper-Middle Income",
        "High Income"
    ],
    duplicates="raise"
)
```

```python
display(
    df["Income_Group"].value_counts(
        dropna=False
    )
)
```

```python
df["Is_Campaign_Responder"] = (
    df["Response"] == 1
).astype(int)
```

```python
validation_results = {
    "Negative Total Spending": (
        df["Total_Spending"] < 0
    ).sum(),

    "Negative Total Purchases": (
        df["Total_Purchases"] < 0
    ).sum(),

    "Negative Total Children": (
        df["Total_Children"] < 0
    ).sum(),

    "Invalid Campaign Count": (
        (df["Total_Accepted_Campaigns"] < 0) |
        (df["Total_Accepted_Campaigns"] > 6)
    ).sum(),

    "Negative Customer Tenure": (
        df["Customer_Tenure_Days"] < 0
    ).sum()
}

validation_df = pd.DataFrame(
    validation_results.items(),
    columns=["Check", "Issue_Count"]
)

display(validation_df)
```
